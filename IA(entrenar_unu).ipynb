```python
import tensorflow as tf
import pandas as pd
import numpy as np
```

```python
from pandas.core.reshape import encoding
#leer el archivo de las temperaturas
grados = pd.read_csv("/content/grados2.csv", sep = ";", encoding= "utf-8")
print(grados)
```

```python
c = grados["c"]
f = grados["f"]

print(c,f)
```

```python
# modelo
modelo = tf.keras.Sequential()
modelo.add(tf.keras.layers.Dense(units=1, input_shape=[1]))
```

```python
# compilar el modelo
modelo.compile(optimizer=tf.keras.optimizers.Adam(1e-3), loss='mean_squared_error')
```

```python
#entrenar el modelo
epocas = modelo.fit(c , f, epochs=200, verbose=1)
```

```python
#prediccion
g = int(input("Celcius: "))
resp = modelo.predict(np.array([g]))
print("Farenheits", resp[0][0])
```
